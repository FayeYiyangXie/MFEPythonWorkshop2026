## 4.9 Data Science: Temperature Conversion

* Define a Fahrenheit function returning the exact result

```python
# fahrenheit function
def fahrenheit(c):
    f = (9 / 5) * c + 32
    return f
```

* For temprature in the range 0-100 degrees, calculate f and round it down to 1 digit of precesion

```python
# create a tabular format
print("Celsius | Fahrenheit")
print("--------|-----------")

# calculate and display the result
for c in range (101):
    f = round(fahrenheit(c),1)
    print(f"{c:7} | {f:10}")
```

## 5.7 Data Science: Duplicate Elimination

* Define a function receiving a list
* For every element from the last to the first, if it is not unique, use `del` to remove it

```python
# define function
def del_dup(test):
    temp = []
    length = len(test)
    n = length - 1
    while n >= 0 :
        if test[n] not in temp:
            temp.append(test[n])
        else:
            del test[n]
        n -= 1
    return test
```

* Test the function with a list of numbers

```python
# define a list of numbers
test_num = [-43, 64.5, 33, 12, -43, 45, 33]
uni_num = del_dup(test_num)
```

* Sort the list

```python
uni_num.sort()
print(uni_num)
```

* Test the function with a list of strings

```python
# define a list of strings
test_str = ['apple', "banana", "Apple", "orange", "banana", "apple"]
uni_str = del_dup(test_str)
```

* Sort the list

```python
uni_str.sort()
print(uni_str)
```

## 5.14 Is a Sequence Sorted?

* Create a function
* Compare every element with the former one, if it's smaller, return False
* Otherwise, proceed this process until we reach the last element

```python
def is_ordered(test):
    length = len(test)
    i = 1
    while i < length:
        if test[i] < test[i - 1]:
            return False
        i += 1
    return True
```

* Use Gen-AI to create test samples

##### Test with lists

```python
print(is_ordered([1, 2, 3, 4, 5]))
print(is_ordered([1, 3, 2, 4, 5]))
print(is_ordered([5, 4, 3, 2, 1]))    
```

##### Test with tuples

```python
print(is_ordered((1, 2, 3, 4)))
print(is_ordered((1, 3, 2, 4)))
print(is_ordered((4, 3, 2, 1)))   
```

##### Test with strings

```python
print(is_ordered('abcde'))
print(is_ordered('abced'))
print(is_ordered('decba'))  
```

## 5.28 Data Science: Survey Response Statistics

* Define the sequence

```python
ratings = [1, 2, 5, 4, 3, 5, 2, 1, 3, 3, 1, 4, 3, 3, 3, 2, 3, 3, 2, 5]
length = 20
```

#### Define ratings

```python
ratings = [1, 2, 5, 4, 3, 5, 2, 1, 3, 3, 1, 4, 3, 3, 3, 2, 3, 3, 2, 5]
length = 20
```

#### Built in functions

##### Min,Max

```python
minimum = min(ratings)
maximum = max(ratings)
```

##### Range

```python
range_ = maximum - minimum
```

##### Mean

```python
mean = sum(ratings)/length
```

##### Median (We already know that n=20, so there's no need to discuss)

```python
sorted_ratings = sorted(ratings)
median = (sorted_ratings[9] + sorted_ratings[10])/2
```

##### Mode

```python
freq = {}
for n in ratings:
    if n in freq:
        freq[n] += 1
    else:
        freq[n] = 1
most_freq = max(freq.values())
mode = []
for n in range(1,6):
    if freq[n] == most_freq:
        mode.append(n)
```

##### Variance

```python
sqr_diff = [(x - mean) ** 2 for x in ratings]
variance = sum(sqr_diff)/length
```

##### Standard Deviation

```python
standard_deviation = variance ** 0.5
```

##### Display Results

```python
print("Minimum:", minimum)
print("Maximum:", maximum)
print("Range:", range_)
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
print("Variance:", variance)
print("Standard Deviation:", standard_deviation)
```
