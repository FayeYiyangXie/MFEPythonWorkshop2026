## 3.10 7% Investment Return

* Firstly, define the original amount invested and the annual rate of return

* Then, use a for loop for the range from year 1 to year 30

````python
for y in range(1,31):
    p = p * (1 + r)
    print('You will have $', round(p,2),' after', y, 'years')
````

## 3.17 Nested Loops

* Recognize that we have 10 rows
* For the first and the second one, we print the given form n times
* For the last two patterns, print `(n-1)` spaces before printing the given form n times
* To display the whole figure, all codes are presented within one block

````python
# display figure (a)
for n in range (1,11):
    print(n * '*', end = '\n')

# separate by one black line
print(15 * '-',end = '\n')

# display figure (b)
for n in range (10,0,-1):
    print(n * '*', end = '\n')

# separate by one black line
print(15 * '-',end = '\n')

# display figure (c)
for n in range (1,11):
    print((n - 1) * ' ',(11 - n) * '*', end = '\n')

# separate by one black line
print(15 * '-',end = '\n')

# display figure (d)
for n in range (10,0,-1):
    print((n - 1) * ' ',(11 - n) * '*', end = '\n')
````

````python
# fig03_03.py
"""Using nested control statements to analyze examination results."""

# initialize variables
passes = 0  # number of passes

# process 10 students
for student in range(10):
    # get one exam result with input validation
    while True:
        try:
            result = int(input('Enter result (1=pass, 2=fail): '))
            if result == 1 or result == 2:
                break  # valid input, exit the loop
            else:
                print('Invalid input. Please enter 1 for pass or 2 for fail.')
        except ValueError:
            print('Invalid input. Please enter a number (1 for pass or 2 for fail).')

    if result == 1:
        passes = passes + 1

# termination phase - calculate failures from total students and passes
failures = 10 - passes
print('Passed:', passes)
print('Failed:', failures)

if passes > 8:
    print('Bonus to instructor')
````
